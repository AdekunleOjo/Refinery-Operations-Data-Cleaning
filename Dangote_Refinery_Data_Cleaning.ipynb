```python
#Treating Outliers
```

```python
#Production Output Data Check
```

```python
production['Actual_Production_Bbl'].describe()
```

```python
Q1 = production['Actual_Production_Bbl'].quantile(0.25)
Q3 = production['Actual_Production_Bbl'].quantile(0.75)

IQR = Q3-Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
```

```python
production['DQ_Issues'] = production['DQ_Issues'].fillna('')

outlier_condition = (production['Actual_Production_Bbl'] < lower_bound) | (production['Actual_Production_Bbl'] > upper_bound)

production.loc[outlier_condition, 'DQ_Issues'] += 'IQR_Outlier|'
production['DQ_Issues'].value_counts()
```

```python
production['DQ_Issues'] = ''
print('Column created/checked')
```

```python
outlier_condition = (production['Actual_Production_Bbl'] < lower_bound) | (production['Actual_Production_Bbl'] > upper_bound)
print(f"Found {outlier_condition.sum()} outliers")
```

```python
Q1 = production['Downtime_Hours'].quantile(0.25)
Q3 = production['Downtime_Hours'].quantile(0.75)

IQR = Q3-Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f"Lower Bound: {lower_bound:.2f}, Upper Bound: {upper_bound:.2f}")
```
